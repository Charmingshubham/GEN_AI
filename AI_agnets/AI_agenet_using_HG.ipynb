```python
import smolagents
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from smolagents import CodeAgent, tool, InferenceClientModel
import requests
import os
```

```python
HF_TOKEN = os.getenv("HUGGINGFACEHUB_API_TOKEN")
```

```python
model = InferenceClientModel(
    model_id='meta-llama/Meta-Llama-3-8B-Instruct',
    api_key=HF_TOKEN
)
```

```python
@tool
def currency_rate(base: str, symbol: str) -> float:
    """
    Fetch currency rate using exchangerate.host API.

    Args:
        base (str): The base currency (e.g., USD).
        symbol (str): The target currency to convert to (e.g., INR).

    Returns:
        float: The exchange rate from base to symbol.
    """
    url = "https://api.exchangerate.host/live?base=USD&symbols=INR"
    params = {
        'access_key':'fec86cb1d9272693b21624146fdb4e6c',
        'base' : 'USD',
        'symbols' : 'INR'}
    response = requests.get(url,params=params)
    data = response.json()
    rate = data['quotes']['USDINR']
    return rate

@tool 
def multiply(a:float,b:float) -> float:
    """
    Multiply two numbers.

    Args:
        a (float): The first number.
        b (float): The second number.

    Returns:
        float: The result of multiplying a and b.
    """
    return a*b
```

```python
agent = CodeAgent(model=model,tools=[multiply,currency_rate])
```

```python
print(agent.run("Convert 50 USD to INR"))
```
