```python
import smolagents
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from smolagents import CodeAgent, Tool, InferenceClientModel
from dotenv import load_dotenv
import os
```

```python
HF_TOKEN = os.getenv("HUGGINGFACEHUB_API_TOKEN")
```

```python
#token=os.getenv('HUGGINGFACEHUB_API_TOKEN')

# we are using InfernceClientModel inplace of huggingfaceendpont because it supports toolcalling and compatible with smolagents to create agent  
model = InferenceClientModel(
    model_id="meta-llama/Meta-Llama-3-8B-Instruct",
    api_key=HF_TOKEN
)
```

```python
# here we have to use tool class cause the @tool will not work with HG models
class Multiply(Tool):
    name = "multiply"
    description = "Multiply two numbers"
    inputs = {
        "a": {"type": "integer", "description": "first number"},
        "b": {"type": "integer", "description": "second number"},
    }
    output_type = "integer"

    def forward(self, a, b):
        return a * b
```

```python
Multiply.name
```

```python
Multiply.description
```

# tool binding

```python
agent = CodeAgent(model=model,tools=[Multiply()])
```

```python
print(agent.run("What is 6 * 9?"))
```
