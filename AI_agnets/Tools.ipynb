# Custom tools

# Method 1 - using @tool

```python
from langchain_core.tools import tool
```

```python
# Step 1 - create a function
def multiply(a,b):
    '''multiply two numbers'''
    return a*b
```

```python
# Step 2 - add type hints
def multiply(a:int,b:int) -> int:
    '''multiply two numbers'''
    return a*b
```

```python
# Step 3 - add tool decorator

@tool
def multiply(a: int, b:int) -> int:
    """multiply two numbers"""
    return a*b
```

```python
result = multiply.invoke({'a':3,'b':4})
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
print(multiply.args_schema.model_json_schema())
```

# Method 2 - Structured tool 

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel , Field
```

```python
class Multiply_input(BaseModel):
    a : int = Field(required=True,discription='the first number to add')
    b : int = Field(required=True,discription='the second number to add')
```

```python
def multiply_fuc(a:int,b:int) -> int:
    return a*b
```

```python
multiply_tool = StructuredTool.from_function(
    func = multiply_fuc,
    name = 'multiply',
    description = 'multiply two numbers',
    args_schema = Multiply_input
)
```

```python
result = multiply_tool.invoke({'a':3, 'b':3})
print(result)
```

```python
print(result)
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

# Method 3 - Using BaseTool Class

```python
from langchain.tools import BaseTool
from typing import Type
```

```python
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
```

```python
class Multiply(BaseTool):
    name : str = 'multiply'
    description : str = 'multiply two numbers'
    args_schema : Type[BaseModel] = MultiplyInput
    def _run(self, a: int, b: int) -> int:
        return a * b
```

```python
multiply_tool = Multiply()
```

```python
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)
```

# ToolKit

```python
from langchain_core.tools import tool

# Custom tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()
```

```python
for tool in tools:
    print(tool.name, "=>", tool.description)
```
